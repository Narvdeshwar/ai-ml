```python
import json
```

```python
# load data
def load_Data(file_name,mode):
    with open(file_name,mode) as f:
        data=json.load(f)
    return data
```

```python
data=load_Data("store_data.json","r")
```

```python
def print_data(data):
    for user in data:
        print(user)
print_data(data)
```

```python
# cleaning of data
## problem encountered
#  1. inconsistent data in rating
#  2. age missing
# 3. duplicate data
```

```python
def clean_data(data):
    # 1. Handle inconsistent rating data
    text_to_num = {
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5
    }

    cleaned_data = []
    unique_user = set()

    for user in data:

        # Normalize rating
        raw_rating = user.get("rating")

        if isinstance(raw_rating, str):
            raw_rating = raw_rating.strip().lower()

            if raw_rating in text_to_num:
                user["rating"] = text_to_num[raw_rating]

        # 2. Handle missing age
        if "age" not in user:
            user["age"] = None

        # 3. Handle duplicate users
        raw_name = user["name"].strip().lower()

        if raw_name not in unique_user:
            unique_user.add(raw_name)
            cleaned_data.append(user)

    return cleaned_data


result = clean_data(data)
```

```python
# clean data after removing the inconsistent data
print_data(result)
```

```python
# recommendation mode
## user rating >4 then product="apple" else product="samsung"
def get_recommendation(data):
    recommendation=[]
    for user in data:
        current_recomm={}
        current_recomm["name"]=user["name"]
        current_recomm["rating"]=user["rating"]
        user_rating=float(user["rating"])
        if user_rating>=4:
            current_recomm["product"]="apple"
        else:
            current_recomm["product"]="samsung"
        recommendation.append(current_recomm)
    return recommendation
data=get_recommendation(result)
print_data(data)
```
