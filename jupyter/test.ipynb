```python
import json
```

```python
# load data
def load_Data(file_name,mode):
    with open(file_name,mode) as f:
        data=json.load(f)
    return data
```

```python
data=load_Data("store_data.json","r")
```

```python
def print_data(data):
    for user in data:
        print(user)
print_data(data)
```

```python
# cleaning of data
## problem encountered
#  1. inconsistent data in rating
#  2. age missing
# 3. duplicate data
```

```python
# 1. inconsistent data in rating
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    for user in data:
        raw_rating=user["rating"]
        if raw_rating in text_to_num:
            user["rating"]=text_to_num[user["rating"].strip().lower()]
clean_data(data)
```
