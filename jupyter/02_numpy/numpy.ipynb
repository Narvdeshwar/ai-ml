```python
# import of numpy
import numpy as np
import time
```

```python
arr=np.array([1,2,3,4,5])
print(arr)
print(type(arr))
```

```python
# execution calculation list vs numpy
size=10_00_00_00
pylist=list(range(size))
start_time=time.time()
sqaure=[x**2 for x in pylist]
end_time=time.time()
print(f"Python list taken time: {end_time-start_time}")

nparray=np.array(pylist)
start_time=time.time()
sqaure=nparray**2
end_time=time.time()
print(f"Numpy array taken time: {end_time-start_time}")

```

```python
# memory
import sys
print(f"Python list size= {sys.getsizeof(pylist)*len(pylist)}")
print(f"numpy size= {nparray.nbytes}")
```
