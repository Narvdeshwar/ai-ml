```python
# import of numpy
import numpy as np
import time
```

```python
arr=np.array([1,2,3,4,5])
print(arr)
print(type(arr))
```

```python
# execution calculation list vs numpy
size=10_00_00_00
pylist=list(range(size))
start_time=time.time()
sqaure=[x**2 for x in pylist]
end_time=time.time()
print(f"Python list taken time: {end_time-start_time}")

nparray=np.array(pylist)
start_time=time.time()
sqaure=nparray**2
end_time=time.time()
print(f"Numpy array taken time: {end_time-start_time}")

```

```python
# memory
import sys
print(f"Python list size= {sys.getsizeof(pylist)*len(pylist)}")
print(f"numpy size= {nparray.nbytes}")
```

```python
# creation of numpy
arr=np.array([1,2,3,4,5])
print(arr)
print(arr.dtype) # data-type of numpy
print(arr.shape) # tells the dimension of numpy
arr2=np.array([12,3,4,5,6,"ravi"])
print(arr2)
print(arr2.dtype) # data-type of numpy
print(arr2.shape) # tells the dimension of numpy
arr3=np.array([[1,2,3],[5,4,5],[1,2,3]])
print(arr3,arr3.shape)
```

```python
arr1=np.zeros((2,2),dtype="int64")
print(arr1,type(arr1),arr1.shape)
arr2=np.ones((3,2)) # by default datatype is float
print(arr2)
# 1d array
arr3=np.ones((5,))
print(arr3,arr3.shape)
```
